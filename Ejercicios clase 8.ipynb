## Ejercicio 2 — Contador de clics

Crea una ventana con un número que comience en cero y un botón **Sumar clic**. Cada clic debe aumentar el número en uno. Agrega otro botón para reiniciar el contador.

**Conceptos:** variables, funciones, `command` y `config()`.

```python
# EJERCICIO 2: escribe aquí tu código
import tkinter as tk

contador = [0] # Se define una variable que empiece en 0

def aumentar_numero():
    contador [0] = contador [0] + 1
    etiqueta.config(text = f"Contador: {contador[0]}") # La f delante de las "" sirve para insertar variables directamente dentro de un texto usando {}, reemplaza el [0] por 1

def reiniciar_contador():
    contador [0] = 0
    etiqueta.config(text = f"Contador: {contador[0]}")

ventana = tk.Tk()
ventana.title("Contador de Clicks")
ventana.geometry("450x300")

etiqueta = tk.Label(
    ventana,
    text="Contador: 0",
    font=("Arial", 16),
    fg="#1B4F72"
)
etiqueta.pack(pady=30)


boton = tk.Button(ventana, text="Sumar Click", command=aumentar_numero)
boton.pack(pady = 10)
boton = tk.Button(ventana, text="Reiniciar", command=reiniciar_contador)
boton.pack(pady = 10)

ventana.mainloop()
```

## Ejercicio 4 — Conversor de guaraníes a dólares

Solicita un monto en guaraníes y una cotización del dólar. Al presionar **Convertir**, calcula y muestra el equivalente en dólares con dos decimales.

Valida que:

- Ningún campo esté vacío.
- Los valores sean numéricos y mayores que cero.
- La cotización no sea cero.

**Conceptos:** formulario con `grid()`, cálculos y validación.

```python
# EJERCICIO 4: escribe aquí tu código, 5900 gs el dolar
import tkinter as tk
from tkinter import messagebox

def convertir(): # Toda la validación
    if entrada1.get() == "" or entrada2.get() == "": # Esto hace que se eviten los campos vacíos.
        messagebox.showerror("Error", "Complete los campos")
        return # Hace que detenga la función para que no siga calculando.

    try: # Se hace para evitar errores al escribir valores que no deben.
        guaranies = float(entrada1.get())
        cotizacion = float(entrada2.get())

        if guaranies <= 0 or cotizacion <= 0: # Se hace dentro del try para evitar valores menores o igual a cero. Sirve como filtro para analizar el valor ingresado.
            messagebox.showerror("Error", "Los valores deben ser mayores que cero")
            return
        # Operación para pasar de gs. a dolares con lo que equivale 1 dolar = 5.900 gs.
        dolares = guaranies / cotizacion # Si el nro es mayor a 0 pasa a realizar la operación.
        
        resultado.config(text = "Resultado: $" + format(dolares, ".2f")) # format(dolares, ".2f(la f es de formato)") sirve para especificar que queremos con 2 nros luego de la coma.

    except ValueError: # Evita valores que no sean números.
        messagebox.showerror("Error", "Ingrese solo números")


ventana = tk.Tk()
ventana.title("Conversor de Gs. a Dólar")
ventana.geometry("450x300")

tk.Label(ventana, text = "Monto en Guaraníes:").grid(row = 0, column = 0, padx = 10, pady = 10, sticky = "e")
entrada1 = tk.Entry(ventana,width = 30)
entrada1.grid(row = 0, column = 1, padx = 10, pady = 10)

tk.Label(ventana, text = "Cotización:").grid(row = 1, column = 0, padx = 10, pady = 10, sticky = "e")
entrada2= tk.Entry(ventana,width = 30)
entrada2.grid(row = 1, column = 1, padx = 10, pady = 10)

tk.Button(ventana, text = "Convertir", command = convertir).grid(row = 2, column = 0, columnspan = 2, pady = 20)

resultado = tk.Label(ventana, text = "Resultado:")
resultado.grid(row = 3, column = 1, padx = 10, pady = 10, sticky = "w")

ventana.mainloop()
```

## Ejercicio 6 — Editor de nota breve

Crea una ventana con un widget `Text` y tres botones:

- **Mostrar:** presenta el contenido en un `messagebox`.
- **Limpiar:** elimina todo el texto.
- **Salir:** solicita confirmación antes de cerrar.

**Conceptos:** `Text`, cuadros de diálogo y eventos.

```python
# EJERCICIO 6: escribe aquí tu código
import tkinter as tk
from tkinter import messagebox

def mostrar():
    contenido = texto.get("1.0", tk.END).strip()
    messagebox.showinfo("Nota", contenido)

def limpiar():
    texto.delete("1.0", tk.END)

def salir():
    if messagebox.askyesno("Salir", "¿Desea cerrar la aplicación?"):
            ventana.destroy()

ventana = tk.Tk()
ventana.title("Editor de Nota breve")
ventana.geometry("450x320")

texto = tk.Text(ventana, width = 40, height = 10)
texto.pack(pady = 10)
mostrar = tk.Button(ventana, text = "Mostrar", command = mostrar)
mostrar.pack(pady = 10)
limpiar = tk.Button(ventana, text = "Limpiar", command = limpiar)
limpiar.pack(pady = 10)
salir = tk.Button(ventana, text = "Salir", command = salir)
salir.pack(pady = 10)

ventana.mainloop()
```

## Ejercicio 8 — Proyecto integrador: registro de visitantes

Construye la aplicación propuesta al inicio de la clase.

### Requisitos mínimos

1. Título y tamaño apropiado.
2. Encabezado con el nombre del sistema.
3. Campos: nombre, documento y motivo de visita.
4. Botones: Registrar, Limpiar y Salir.
5. Validación de campos obligatorios.
6. Confirmación mediante `messagebox`.
7. Limpieza automática después de registrar.
8. Cursor colocado nuevamente en el campo nombre.
9. Tecla Enter asociada a la acción Registrar.
10. Organización mediante `Frame` y `grid()`.

### Desafío adicional

Guarda temporalmente cada registro en una lista y muestra en una etiqueta cuántos visitantes fueron registrados.

```python
# EJERCICIO 8 - PROYECTO INTEGRADOR
# Construye aquí el registro de visitantes.
import tkinter as tk
from tkinter import messagebox

ventana = tk.Tk()
ventana.title("Registro de Visitantes")
ventana.geometry("500x400")

encabezado = tk.Frame(ventana, bg="#1F618D")
encabezado.pack(fill="x") # fill = "x" sirve para rellenar el color del encabezado de forma horizontal toda la ventana.
tk.Label(encabezado, text="REGISTRO DE VISITANTES", bg="#1F618D", fg="white",
         font=("Arial", 15, "bold")).pack(pady=15)

formulario = tk.Frame(ventana)
formulario.pack(pady = 20)

tk.Label(formulario, text="Nombre:").grid(row=0, column=0, padx = 10, pady = 10)
entrada_nombre = tk.Entry(formulario, width=30)
entrada_nombre.grid(row=0, column=1, padx=10, pady=10)

tk.Label(formulario, text="Documento:").grid(row=1, column=0, padx = 10, pady=10)
entrada_documento = tk.Entry(formulario, width=30)
entrada_documento.grid(row=1, column=1, padx=10, pady=10)

tk.Label(formulario, text = "Motivo de Visita:").grid(row=2, column=0, padx=10, pady=10)
entrada_motivo = tk.Entry(formulario, width = 30)
entrada_motivo.grid(row=2, column=1, padx=10, pady=10)

cantidad = tk.Label(formulario, text = "Visitantes registrados: 0")
cantidad.grid(row = 5, column = 0, columnspan = 2, pady = 10)

visitantes = []

def registrar():
    nombre = entrada_nombre.get()
    documento = entrada_documento.get()
    motivo = entrada_motivo.get()

    if nombre == "" or documento == "" or motivo == "":
        messagebox.showerror("Error", "Complete todos los campos")
        return

    visitantes.append([nombre, documento, motivo]) # Para guardar los datos en la lista visitantes = []

    cantidad.config(text = "Visitantes registrados: " + str(len(visitantes)))

    messagebox.showinfo("Registro", "Visitante registrado correctamente")

    limpiar()
    entrada_nombre.focus()

def registrar_enter(event):
    registrar()

def limpiar():
    entrada_nombre.delete(0, tk.END)
    entrada_documento.delete(0, tk.END)
    entrada_motivo.delete(0, tk.END)
    entrada_nombre.focus()

def salir():
    respuesta = messagebox.askyesno("Salir", "¿Desea salir?")

    if respuesta:
        ventana.destroy()

tk.Button(formulario, text = "Registrar", command = registrar).grid(row = 3, column = 0, padx = 10, pady = 15)
tk.Button(formulario, text = "Limpiar", command = limpiar).grid(row = 3, column = 1, padx = 10, pady = 15)
tk.Button(formulario, text = "Salir", command = salir).grid(row = 4, column = 0, columnspan = 2, pady = 10)


ventana.bind("<Return>", registrar_enter)
ventana.mainloop()
```
