```python
# Ejercicio 5

import tkinter as tk
from tkinter import messagebox, ttk

ventana = tk.Tk()
ventana.title("Buscar Producto")
ventana.geometry("420x300")

def buscar():
    texto = (entrada.get())

    coincidencias = 0

    for producto in productos:
        if texto.lower() == producto.lower():
            coincidencias += 1

    messagebox.showinfo("Resultado", "Coincidencias encontradas: " + str(coincidencias))

def buscar_enter(event):
    buscar()


productos = ["Harina", "Vaso", "Silla", "Termómetro", "Silla"]

ttk.Label(ventana, text = "Buscar producto:").pack(pady = 15)
entrada = ttk.Entry(ventana)
entrada.pack()

ttk.Button(ventana, text = "Buscar", command = buscar).pack(pady = 20)

entrada.bind("<Return>", buscar_enter)
ventana.mainloop()
```

```python
# Reto integrador - Bloc de notas mejorado

import tkinter as tk
from tkinter import ttk, messagebox, filedialog

archivo_actual = None

ventana = tk.Tk()
ventana.title("Bloc de notas mejorado")
ventana.geometry("480x400")

def mostrar():
    contenido = texto.get("1.0", tk.END).strip()
    messagebox.showinfo("Nota", contenido)

def limpiar():
    texto.delete("1.0", tk.END)

def confirmar_guardado():
    contenido = texto.get("1.0", tk.END).strip()

    if contenido == "":
        return True
    
    respuesta = messagebox.askyesnocancel("Guardar cambios", "Desea guardar los cambios?")

    if respuesta is True:
        return guardar()
    elif respuesta is False:
        return True
    else:
        return False

def nuevo():
    global archivo_actual

    if confirmar_guardado():
        texto.delete("1.0", tk.END)
        archivo_actual = None
        estado.config(text = "Nuevo documento")

def abrir():
    global archivo_actual

    if not confirmar_guardado():
        return

    ruta = filedialog.askopenfilename(filetypes = [("Archivos de texto", "*.txt"), ("Todos los archivos", "*.*")])

    if ruta:
        try:
            with open(ruta, "r", encoding = "utf-8") as archivo:
                contenido = archivo.read()
            texto.delete("1.0", tk.END)
            texto.insert("1.0", contenido)
            archivo_actual = ruta

            estado.config(text = "Archivo abierto")

        except:
            messagebox.showerror("Error", "No se pudo abrir el archivo")

def guardar():
    global archivo_actual

    if archivo_actual is None:
        return guardar_como()
    try:
        contenido = texto.get("1.0", tk.END).strip()

        with open(archivo_actual, "w", encoding = "utf-8") as archivo:

            archivo.write(contenido)
            estado.config(text = "Archivo guardado")
            return True
    except:
        messagebox.showerror("Error", "No se pudo guardar el archivo")
        return False

def guardar_como():
    global archivo_actual
    
    ruta = filedialog.asksaveasfilename(defaultextension = ".txt", filetypes = [("Archivos de texto", "*.txt"), ("Todos los archivos", "*.*")])
    if ruta:
        archivo_actual = ruta
        return guardar()
    return False

def salir():
    if confirmar_guardado():
        ventana.destroy()

def contar_palabras():
     contenido = texto.get("1.0", tk.END)
     cantidad = len(contenido.split())
     estado.config(text = "Palabras: " + str(cantidad))


texto = tk.Text(ventana, width = 40, height = 10)
texto.pack(pady = 10)

barra_menu = tk.Menu(ventana)
menu_archivo = tk.Menu(barra_menu, tearoff = 0)
menu_archivo.add_command(label="Nuevo", command = nuevo)
menu_archivo.add_command(label="Abrir", command = abrir)
menu_archivo.add_command(label="Guardar", command = guardar)
menu_archivo.add_command(label="Guardar como...", command = guardar_como)
menu_archivo.add_separator()
menu_archivo.add_command(label="Salir", command = salir)
barra_menu.add_cascade(label="Archivo", menu=menu_archivo)

menu_opciones = tk.Menu(barra_menu, tearoff = 0)
menu_opciones.add_command(label="Contar palabras", command = contar_palabras)
barra_menu.add_cascade(label = "Opciones", menu = menu_opciones)

ventana.config(menu=barra_menu)

mostrar = ttk.Button(ventana, text = "Mostrar", command = mostrar)
mostrar.pack(pady = 10)

limpiar = ttk.Button(ventana, text = "Limpiar", command = limpiar)
limpiar.pack(pady = 10)

salir = ttk.Button(ventana, text = "Salir", command = salir)
salir.pack(pady = 10)

estado = ttk.Label(ventana, text = "Listo")
estado.pack(fill = "x")

ventana.mainloop()
```
